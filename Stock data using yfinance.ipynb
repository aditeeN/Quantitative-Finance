```python
import yfinance as yf
import numpy as np
```

```python
stock = "MSFT"

# Create a Ticker object
ticker = yf.Ticker(stock)
print(ticker)

# Fetch historical market data for the last 30 days
historical_data = ticker.history(period="1mo")  # data for the last month

# Display a summary of the fetched data
print(f"Summary of Historical Data for {stock}:")
print(historical_data[['Open', 'High', 'Low', 'Close', 'Volume']])
```

```python
historical_data['Close'].describe()
```

```python
historical_data['returns'] = np.log(historical_data['Close']/historical_data['Close'].shift(1))
historical_data['returns']*100
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize = (12, 3))
ax.plot(historical_data['returns']*100)
ax.axhline(historical_data['returns'].mean(), color = 'r')
plt.show()
```
