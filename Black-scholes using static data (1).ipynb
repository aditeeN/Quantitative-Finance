```python
import math
import numpy as np
```

```python
from scipy.stats import norm 
```

```python
s = 45
k = 40
t = 2
r = 0.1
vol = 0.5
```

```python
d1= (math.log(s/k)+(r+ 0.5* (vol**2))*t)/(vol*math.sqrt(t))
d1
```

```python
d2 = d1 - (vol*math.sqrt(t))
d2
```

```python
c = s * norm.cdf(d1) - k*math.exp(-r*t)*norm.cdf(d2)
c
```

```python
p = k*math.exp(-r*t)*norm.cdf(-d2) - s*norm.cdf(-d1)
p
```
