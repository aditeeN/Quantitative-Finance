```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup as bs
import json
```

```python
def get_bond_data(url, parser, classname):
    r = requests.get(str(url))
    data = bs(r.text,str(parser))
    data = data.find('table', attrs={"class":str(classname)})
    ttr = data.find_all('tr')
    tlist = []
    for h in ttr:
        tlist.append([i.get_text(strip=True) for i in h.find_all(['td','th']) if i.get_text(strip = True)])
    tt = tlist
    #tt = getrowsfromtable(data)
    df = pd.DataFrame(tt[2:])
    index = df[0]
    df = df.drop(df.columns[[0,9,10]], axis = 1)
    a = pd.DataFrame([np.repeat(tt[0][1:3],4), tt[1]]).T
    #header = pd.MultiIndex.from_frame(a)
    df.columns = a
    df[tt[0][0]] = index
    df = df.set_index(tt[0][0])
    return df
```
